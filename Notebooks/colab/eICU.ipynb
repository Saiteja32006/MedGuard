```python
import pandas as pd
EXPECT = {
 "patient":    ["patientunitstayid","hospitalid","gender","age","unittype"],
 "medication": ["patientunitstayid","drugname","dosage","routeadmin",
                "drugstartoffset","drugordercancelled"],
 "lab":        ["patientunitstayid","labname","labresult","labresultoffset"],
 "allergy":    ["patientunitstayid","allergyname","allergytype","allergyoffset"],
}
for t, cols in EXPECT.items():
    df = pd.read_csv(f"/content/eicu/{t}.csv.gz", nrows=5)
    missing = [c for c in cols if c not in df.columns]
    print(f"{t:<12}", "OK" if not missing else f"MISSING {missing}")
    if missing: print("   actual:", list(df.columns))
```

```python
import pandas as pd, collections
c = collections.Counter()
for ch in pd.read_csv("/content/eicu/lab.csv.gz", usecols=["labname"],
                      chunksize=2_000_000):
    c.update(ch.labname.dropna().values.tolist())

for m in ["creatinine","potassium","PT - INR","ALT (SGPT)","glucose",
          "bedside glucose","platelets x 1000","WBC x 1000","TSH"]:
    print(f"  {'OK  ' if c.get(m) else 'MISS'} {m:<20} {c.get(m,0):>10,}")
print("\ntop 30 actual:")
for k, v in c.most_common(30): print(f"  {v:>10,}  {k}")
```
